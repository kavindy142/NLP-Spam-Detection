# SVM Model Training

This notebook trains a Support Vector Machine (SVM) classifier using the TF-IDF features generated in the previous notebook.

Steps:
1. Import libraries
2. Load preprocessed dataset
3. Split dataset
4. Apply TF-IDF
5. Train SVM
6. Make predictions

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
```

```python
df = pd.read_csv("../data/preprocessed_sms_spam.csv")

print(df.shape)

df.head()
```

```python
text_column = "clean_message"
target_column = "label_encoded"

X = df[text_column].fillna("")
y = df[target_column]

print("Messages:", len(X))
```

```python
X_train_text, X_test_text, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(X_train_text.shape)
print(X_test_text.shape)
```

```python
tfidf_vectorizer = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1,2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

X_train = tfidf_vectorizer.fit_transform(X_train_text)
X_test = tfidf_vectorizer.transform(X_test_text)

print(X_train.shape)
print(X_test.shape)
```

```python
svm_model = SVC(
    kernel="linear",
    C=1.0,
    random_state=42
)

svm_model.fit(X_train, y_train)

print("SVM model trained successfully!")
```

```python
y_pred = svm_model.predict(X_test)

print("Prediction completed.")
```

```python
results = pd.DataFrame({
    "Message": X_test_text.reset_index(drop=True),
    "Actual": y_test.reset_index(drop=True),
    "Predicted": y_pred
})

results.head(10)
```

```python
import joblib
import os

os.makedirs("../models", exist_ok=True)

joblib.dump(svm_model, "../models/svm_model.pkl")

print("SVM model saved successfully!")
```
