```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
PROJECT_ROOT = os.path.abspath("..")

RESULTS_DIR = os.path.join(PROJECT_ROOT, "results")
```

```python
svm_metrics = pd.read_csv(
    os.path.join(RESULTS_DIR, "svm_evaluation_metrics.csv")
)

gru_metrics = pd.read_csv(
    os.path.join(RESULTS_DIR, "gru_evaluation_metrics.csv")
)

print("Evaluation files loaded now")
```

```python
comparison = pd.concat(

    [svm_metrics, gru_metrics],

    ignore_index=True

)

comparison
```

```python
comparison.to_csv(

    os.path.join(

        RESULTS_DIR,

        "model_comparison.csv"

    ),

    index=False

)

print("Comparison saved.")
```

```python
plt.figure(figsize=(6,4))

plt.bar(

    comparison["Model"],

    comparison["Accuracy"]

)

plt.title("Model Accuracy Comparison")

plt.ylabel("Accuracy")

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.bar(

    comparison["Model"],

    comparison["Precision"]

)

plt.title("Precision Comparison")

plt.ylabel("Precision")

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.bar(

    comparison["Model"],

    comparison["Recall"]

)

plt.title("Recall Comparison")

plt.ylabel("Recall")

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.bar(

    comparison["Model"],

    comparison["F1_Score"]

)

plt.title("F1 Score Comparison")

plt.ylabel("F1 Score")

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.bar(

    comparison["Model"],

    comparison["ROC_AUC"]

)

plt.title("ROC-AUC Comparison")

plt.ylabel("ROC-AUC")

plt.grid(axis="y")

plt.show()
```

```python
best_accuracy = comparison.loc[

    comparison["Accuracy"].idxmax()

]

print("Best model (based on Accuracy)")

print(best_accuracy)
```
